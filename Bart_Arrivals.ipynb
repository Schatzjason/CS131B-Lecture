```python
import pandas as pd

bart_url = "https://raw.githubusercontent.com/Schatzjason/CS131B-Lecture/refs/heads/main/bart_arrivals.csv"

df = pd.read_csv(
    bart_url,
    parse_dates=["expected_arrival", "actual_arrival"]
    )

df["delay"] = (
    (df["actual_arrival"] - df["expected_arrival"]).dt.total_seconds()
)

df.dtypes


```
